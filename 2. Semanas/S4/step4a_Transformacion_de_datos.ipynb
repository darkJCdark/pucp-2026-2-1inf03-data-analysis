## Estandarización, Z score: (X - X_mean) / X_std

(X - X_mean) / X_std

```python
datos.describe()
```

```python
def z_score(x):
  mu = x.mean()
  sigma = x.std()
  z_score = (x - mu) / sigma
  return z_score, mu, sigma
```

```python
datos['age_standarized_v1'], age_mean, age_std = z_score(datos.age)
```

```python
datos.age.values
```

```python
datos.age.values.reshape(-1, 1)
```

```python
from sklearn.preprocessing import StandardScaler
StandardScaler().fit_transform(datos.age.values.reshape(-1, 1))
datos['age_standarized_v2'] = StandardScaler().fit_transform(datos.age.values.reshape(-1, 1))
datos['age_standarized_v2']
```

## Normalización o Escalamiento: (X - X_min) / (X_max - X_min)

(X - X_min) / (X_max - X_min)

```python
datos.describe()
```

```python
def min_max(x):
  xmin = x.min()
  xmax = x.max()
  min_max = (x - xmin) / (xmax - xmin)
  return min_max, xmin, xmax
```

```python
datos['age_scaled'], age_min, age_max = min_max(datos.age)
datos['age_scaled']
```

```python
datos['age_scaled'].max()
```

```python
datos['age_scaled'].min()
```

Otra opción de Min-Max-Scaler: https://scikit-learn.org/stable/modules/generated/sklearn.preprocessing.MinMaxScaler.html

```python
print('Min:', age_min)
print('Max:', age_max)

datos[['age', 'age_scaled']].sample(5)
```
