# **Caso 1 -> Reducción de dimensiones: Áreas de actividad económica**

Los datos sobre el Producto Geográfico Bruto (GSP) de los estados de EEUU están disponibles en el archivo **datosGSP.csv** (en millones de dólares). El archivo consiste de 50 filas (una por cada estado) y 14 columnas, la primera contiene el ID del estado y las 13 restantes corresponden a las siguientes áreas de la actividad económica:

* (V2) Agricultura, silvicultura y pesquería
* (V3) Minería
* (V4) Construcción
* (V5) Manufactura (bienes de consumo duradero)
* (V6) Manufactura (bienes no duraderos)
* (V7) Transporte
* (V8) Comunicaciones
* (V9) Electricidad, gas y saneamiento.
* (V10) Comercio mayorista
* (V11) Comercio minorista
* (V12) Fiduciaria, bienes raíces y seguros
* (V13) Servicios
* (V14) Gobierno

*GSP = Gross State Product*

*Gross: se refiere a "bruto", es decir, el valor total sin deducir impuestos o depreciación*

*State: hace referencia al estado (en EE.UU., las divisiones políticas son los "estados")*

*Product: hace referencia a la producción de bienes y servicios*
